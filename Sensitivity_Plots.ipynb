```python
import pandas as pd
import numpy as np
 
path = r"Final_Data.csv"

df = pd.read_csv(path)
df.head()
```

```python
df['Ethylene con: (ppm)'] = np.exp(df['Ethylene con: (ppm)'])
```

```python
import matplotlib.pyplot as plt

# Plot two columns
plt.figure(figsize=(10, 6))
plt.scatter(df['Ethylene con: (ppm)'], df['TGS2620-3'], c='blue', marker='o', edgecolors='black', alpha=0.7)
plt.xlabel('ETHYLENE CONCENTRATION (PPM)')
plt.ylabel('SENSITIVITY (MIN-MAX NORMALISED READINGS)')
plt.title('SENSITIVITY OF TGS2620-3 SENSOR TO ETHYLENE CONCENTRATION')
plt.grid(True)
plt.show()
```
